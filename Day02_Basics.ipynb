```python
# Operators and its type

# Arithmetic
num_1 = 20
num_2 = 20

# "+" operator -> ADD

print(num_1 + num_2)

# "-" operator -> SUB

print(num_1 - num_2)

# "*" operator

print(num_1 * num_2)

# "/" operator -> The result will always be in floating point(Decimals)

print(num_1 / num_2)


```

```python
# "%" Modulus operator

print(5 % 3)
```

```python
# "//" double division, it only returns the value in integer and not the decimal

num_1 = 10
num_2 = 3

result = num_1 // num_2

print(result)
```

```python
# taking the power/exponential

print(2**3.5)
```

```python
# Relational

# > = (a > b) -> if a is greater than b
print(5>6)

# <

print(5<6)

# >=
print(6>=6)

# <=
print(-3.5>=-2)
# ==

print(2.9999999 == 3)

# !=

print(2.9999999 != 3)
```

```python
# Logical
# AND

# print((4 > 5) and ( 5> 6))


# OR

# print((4 < 5) or ( 5> 6))
# NOT

print(not(4!=5))
```

```python
# Bitwise

# Bitwise AND -> &

print(3 & 3)

# Bitwise OR -> |

print(1 | 2)

# Bitwise NOT -> ~

print(~3)

# Bitwise XOR -> ^

print(3 ^ 2)

# Bitwise shift left

print(3 << 2)

# Bitwise shift right

print(5 >> 2)

```
